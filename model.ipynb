```python
import pandas as pd
import joblib
from sklearn.svm import LinearSVC
from sentence_transformers import SentenceTransformer

print("Бібліотеки успішно завантажено!")
```

```python
# 1. Навчальні дані для 5 інтентів ShelterED
texts = [
    # --- explain (пояснення теорії, правил) ---
    "поясни правило",
    "що таке фотосинтез",
    "не розумію цю тему з фізики",
    "розкажи простими словами про дроби",
    "як працює цикл for в python",
    "що означає цей термін",
    "не можу зрозуміти закон ома",
    "поясни будь ласка теорему піфагора",

    # --- practice (практика, задачі, приклади) ---
    "дай задачу",
    "хочу попрактикуватись",
    "як це розв'язати",
    "накинь приклад на практику",
    "дай якесь завдання по програмуванню",
    "хочу закріпити тему вправами",
    "підкинь задачу на цикли",

    # --- check (перевірка рішень учня) ---
    "перевір мою відповідь",
    "це правильно?",
    "ось що в мене вийшло, подивись",
    "чи вірне таке розв'язання",
    "перевір код чи є тут помилки",
    "скажи чи я правильно порахував",

    # --- support (психологічна підтримка, вигорання) ---
    "я втомився",
    "мені тривожно перед іспитом",
    "не можу більше, здаюсь",
    "паніка, нічого не встигаю",
    "я занадто тупий для цього",
    "у мене вигорання, не хочу нічого робити",
    "мені сумно і важко",

    # --- chat (загальні розмови, цікаві факти) ---
    "а чому чорні діри такі масивні",
    "як працює штучний інтелект",
    "розкажи цікавий факт про космос",
    "що буде якщо впасти в чорну діру",
    "як винайшли комп'ютер",
    "який сьогодні день тижня"
]

labels = [
    "explain", "explain", "explain", "explain", "explain", "explain", "explain", "explain",
    "practice", "practice", "practice", "practice", "practice", "practice", "practice",
    "check", "check", "check", "check", "check", "check",
    "support", "support", "support", "support", "support", "support", "support",
    "chat", "chat", "chat", "chat", "chat", "chat"
]

```

```python
embedder = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

print("Генерація векторів...")
X_embeddings = embedder.encode(texts)


print("Навчання LinearSVC...")
classifier = LinearSVC(random_state=42)
classifier.fit(X_embeddings, labels)

print("Модель успішно натренована!")
```

```python
# Розширений тест: складніші формулювання на тему шкільної освіти без прямих ключових слів
extended_test_queries = [
    # -- Очікуємо: explain (учень не розуміє суть теми, просить розкрити концепт) --
    "Я взагалі не курю, як ця тригонометрія працює у реальному житті",
    "Мені важко в'їхати в цю тему з біології про клітини",
    "Розкажи на пальцях, чому виникає електричний струм",
    "Кіногерої весь час говорять про квантову фізику, що це за звір",

    # -- Очікуємо: practice (учень хоче тренуватись, але без слова "задача") --
    "Шкода, що в підручнику мало прикладів для набиття руки",
    "Хотів би спробувати самостійно порахувати щось подібне",
    "Давай влаштуємо невеликий інтерактив на закріплення",
    "Є якісь тренажери, щоб набити руку на рівняннях?",

    # -- Очікуємо: check (учень показує результат або просить оцінити) --
    "Глянь оком на моє домашнє завдання, чим чорт не жартує",
    "Чи правильно я тут виписав винятки з правил?",
    "Ось моє творіння з літератури, що скажеш про якість?",
    "Чи є тут логічні дірки в моєму розв'язанні?",

    # -- Очікуємо: support (емоції, стрес, тиск школи, вигорання) --
    "Цей НМТ мене доконає, я вже не вивожу стільки інфи",
    "Батьки тиснуть через оцінки, а я більше не можу дивитися на ці тести",
    "Руки опускаються, нічого не запам'ятовується, відчуваю себе повним нулем",
    "Сиджу о третій ночі над конспектами і плакати хочеться від безсилля",

    # -- Очікуємо: chat (відволікання, навколоосвітні чи наукові цікавинки) --
    "Цікаво, чому в школах досі не вивчають космос глибше?",
    "Як думаєш, чи замінять нас повністю алгоритми в майбутньому?",
    "Яка ймовірність того, що ми знайдемо розумне життя поряд?",
    "Розкажи якийсь дивний історичний факт, про який мовчать на уроках"
]

# Проганяємо через нашу модель
test_embeddings = embedder.encode(extended_test_queries)
predicted_intents = classifier.predict(test_embeddings)

# Виводимо результати з форматуванням
print(f"{'ФРАЗА УЧНЯ':<65} | ІНТЕНТ")
print("-" * 80)
for q, intent in zip(extended_test_queries, predicted_intents):
    print(f"«{q}»".ljust(67) + f"➔ **{intent}**")
```
